```python
#5-sided spinner using uniform
import numpy as np

x = np.random.uniform(0, 1)

if x < 1/5:
    spin = 1
elif x < 2/5:
    spin = 2
elif x < 3/5:
    spin = 3
elif x < 4/5:
    spin = 4
else:
    spin = 5

spin
```

```python
#roll 4-sided die until sum>=15

total_sum = 0
num_rolls = 0

while total_sum < 15:
    roll = np.random.randint(1, 5)  # 1,2,3,4
    total_sum += roll
    num_rolls += 1

num_rolls, total_sum
```

```python
#stop when two identical rolls in a row

num_rolls = 0

prev = np.random.randint(1, 5)
num_rolls += 1

while True:
    curr = np.random.randint(1, 5)
    num_rolls += 1

    if curr == prev:
        break

    prev = curr

num_rolls
```

```python
# sum of squares up to a random n
n = np.random.randint(5, 51)  # 5 to 50 inclusive
sum_squares = 0

for k in range(1, n + 1):
    sum_squares += k**2

n, sum_squares
```

```python
#replace invalid humidity values with none
humidity_data = [45, 50, -10, 60, 200, 55, 48, 52]

for i in range(len(humidity_data)):
    if humidity_data[i] < 0 or humidity_data[i] > 100:
        humidity_data[i] = None

humidity_data
```

```python
#average values
#ignore invalid values
humidity_data = [45, 50, None, 60, None, 55, 48, 52]

total = 0
count = 0

for val in humidity_data:
    if val is not None:
        total += val
        count += 1

avg1 = total / count
avg1

#stop at first invalud, average valid before it
humidity_data_raw = [45, 50, -10, 60, 200, 55, 48, 52]

total = 0
count = 0

for val in humidity_data_raw:
    if val < 0 or val > 100:
        print("Warning: invalid humidity detected")
        break
    total += val
    count += 1

avg2 = total / count
avg2
```

```python
#count 4s before first 2

num_fours = 0

while True:
    roll = np.random.randint(1, 5)
    if roll == 2:
        break
    if roll == 4:
        num_fours += 1

num_fours
```

```python
#stop when 2 of last 3 rolls are 1

rolls = []
num_rolls = 0

while True:
    roll = np.random.randint(1, 5)
    rolls.append(roll)
    num_rolls += 1

    if len(rolls) >= 3:
        last3 = rolls[-3:]
        if last3.count(1) >= 2:
            break

num_rolls, rolls[-3:]
```
